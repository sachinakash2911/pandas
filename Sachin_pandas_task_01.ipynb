```python
today = pd.Timestamp.today()
```

```python
df['age'] = today.year - df['birth_date'].dt.year
```

```python
df['age'] = today.year - df['birth_date'].dt.year

df.loc[
    (df['birth_date'].dt.month > today.month) |
    (
        (df['birth_date'].dt.month == today.month) &
        (df['birth_date'].dt.day > today.day)
    ),
    'age'
] -= 1
```
